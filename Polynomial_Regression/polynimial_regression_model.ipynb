## Polynomial Regression Model

### Import Libraries 

```python
# Step 1: Import the required libraraies

import pandas as pd 
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Step 2: Load the datasets 
df = pd.read_csv("polynominal_dataset.csv")
```

```python
# Step 3: EDA

plt.scatter(df["Temperature (°C)"],df["Ice Cream Sales (units)"])

plt.title("Polt the data")
plt.xlabel("Temperature")
plt.ylabel("Ice Cream Sales")
```

```python
# Step 4: Divide the data into X and y 


X = df["Temperature (°C)"].values.reshape(-1,1)
y = df["Ice Cream Sales (units)"]

```

```python
# polynomial regression

from sklearn.preprocessing import PolynomialFeatures

# create the instance 
poly = PolynomialFeatures(degree=2)

# fit and transform the d
x_poly = poly.fit_transform(X)
```

```python
x_poly
```

```python
# Step 5: Model building

from sklearn.linear_model import LinearRegression

# create the instance of model 
LR = LinearRegression()

# Train the modle on the X_poly

model = LR.fit(x_poly,y)

y_pred = model.predict(x_poly)

```

```python
# model evaluation metrics

from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score

# find the MSE
MSE = mean_squared_error(y,y_pred)

# find the MAE
MAE =  mean_absolute_error(y,y_pred)

# find the r2 score
r2_score = r2_score(y,y_pred)

# print all metrics
print(f"MSE : {MSE}")
print(f"MAE : {MAE}")
print(f"R2_Score : {r2_score}")
```
